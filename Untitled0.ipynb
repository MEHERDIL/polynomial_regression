```python
import numpy as np
import matplotlib.pyplot as plt

# Internuclear distance
x = np.linspace(-2, 2, 1000)

# Harmonic oscillator potential (molecular vibration model)
V = x**2

# Vibrational energy levels
energy_levels = [0.5, 1.5, 2.5, 3.5]

plt.figure(figsize=(8,6))

# Plot potential energy curve
plt.plot(x, V, color='black', linewidth=2, label='Potential Energy Curve')

# Plot vibrational energy levels
for n, E in enumerate(energy_levels):
    plt.hlines(E, -1.5, 1.5, linestyles='dashed')

    # Vibrational wave-like shape on each level
    wave = 0.15*np.sin((n+1)*5*x) + E
    plt.plot(x, wave)

# Labels
plt.xlabel("Internuclear Distance")
plt.ylabel("Energy")
plt.title("Vibrational Energy Levels in a Molecule")
plt.legend()

plt.show()
```
